```python
# 📚 Gerekli Kütüphaneler
import os
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# ✅ Chain sistemleri yeni modüle taşındı
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.documents import Document
from langchain_core.output_parsers import StrOutputParser

# 🤗 Hugging Face pipeline
from langchain_huggingface import HuggingFacePipeline
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
```

```python
# 1️⃣ Veriyi Yükleme ve Parçalama
print("Veri yükleniyor ve parçalanıyor...")

# Dosyayı yükle
loader = TextLoader("anayasa.txt", encoding="utf-8")
documents = loader.load()  # documents = list of Document objeleri


# Metni parçalara böl
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)


docs = text_splitter.split_documents(documents)  # docs = list of Document objeleri

print(f"Metin toplam {len(documents)} belgeden {len(docs)} parçaya bölündü.")
for i, doc in enumerate(docs[:3], 1):  # İlk 3 parçayı göster
    print(f"\n--- Parça {i} ---\n{doc.page_content[:500]}...")  # İlk 500 karakter
```

```python
# 2️⃣ Embedding ve Vektör Veritabanı Oluşturma
print("\nEmbedding modeli yükleniyor ve vektör veritabanı oluşturuluyor...")

# HuggingFace embeddings
embeddings = HuggingFaceEmbeddings(
    # model_name="sentence-transformers/paraphrase-multilingual-mpnet-base-v2", # <-- Yeterince iyi değil
    model_name="BAAI/bge-m3",  # <-- En yeni (SOTA) embedding modeli
    model_kwargs={"device": "cpu"}
    # bge-m3 için "encode_kwargs={'normalize_embeddings': True}" gerekebilir,
    # ama önce böyle deneyin.
)

# FAISS vektör veritabanı oluştur
print("Embedding modeli yükleniyor... (bge-m3 büyüktür, 2.3GB, zaman alabilir)")
vector_store = FAISS.from_documents(docs, embeddings)

print("Vektör veritabanı başarıyla oluşturuldu.")
```

```python
# 3️⃣ Cevap Üretecek Google Gemini Modelini Kurma
print("\nCevap üretici LLM (Google Gemini) kuruluyor...")

import google.generativeai as genai
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata

# API anahtarını Colab Secrets'tan güvenle al
try:
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=GOOGLE_API_KEY)
    print("Google API anahtarı başarıyla yüklendi.")
except userdata.SecretNotFoundError:
    print("HATA: 'GOOGLE_API_KEY' Colab Secrets'ta bulunamadı.")
    # Hata durumunda llm'i None olarak ayarlayabilirsiniz
    llm = None
except Exception as e:
    print(f"API anahtarı yüklenirken bir hata oluştu: {e}")
    llm = None

# LangChain için Gemini modelini kur
# RAG için 'gemini-1.5-flash' hızlı ve etkilidir
if 'GOOGLE_API_KEY' in locals():
    llm = ChatGoogleGenerativeAI(
        model="models/gemini-2.5-flash", # <-- Listeden gelen DOĞRU AD
        temperature=0
        )
    print("Google Gemini (models/gemini-2.5-flash) LLM başarıyla kuruldu.")
```

```python
# 4️⃣ RAG Zincirini Kurma ve Çalıştırma
print("\nRAG zinciri kuruluyor ve test ediliyor...")

from langchain_core.runnables import RunnableSequence
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# --- Prompt şablonu
prompt_template = "Aşağıdaki metinleri kullanarak soruyu cevapla:\n\n{context}\n\nSoru: {question}\nCevap:"
prompt = ChatPromptTemplate.from_template(prompt_template)

# --- Zincir tanımı (doğru kullanım)
chain = RunnableSequence(
    prompt,    # Önce prompt ile context ve question hazırlanacak
    llm,       # HuggingFace LLM ile cevap üretilecek
    StrOutputParser()  # Sonucu düz metin olarak al
)

# ... (Hücre 4'ün başı)

# --- Sorgu ve ilgili belgeleri FAISS'ten çek
query = "Vatandaşların seçme ve seçilme hakları"
relevant_docs = vector_store.similarity_search(query, k=5)

# --- Belgeleri zincire uygun formata çevir
context_text = "\n\n".join([doc.page_content for doc in relevant_docs])
inputs = {"context": context_text, "question": query}


# HATA AYIKLAMA: LLM'e giden bağlamı yazdır
# ----------------------------------------------------
print("\n" + "*"*30)
print("--- LLM'E GÖNDERİLEN BAĞLAM (CONTEXT) ---")
print(context_text)
print("*"*30 + "\n")
# ----------------------------------------------------


# --- Cevap üret
response = chain.invoke(inputs)

# --- Sonucu Yazdır ---
print("\n" + "="*50)
print("--- SORU ---")
print(query)
print("\n--- CEVAP (LLM'den gelen) ---")
print(response)  # <-- BU SATIR ÇOK ÖNEMLİ
print("="*50)

# ... (Hücre 4'ün sonu)
```
